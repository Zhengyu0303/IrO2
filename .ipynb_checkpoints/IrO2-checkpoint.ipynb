```julia
using Plots
using DelimitedFiles
```

```julia
data = readdlm("/Users/zhengyuliu/Library/CloudStorage/OneDrive-ImperialCollegeLondon/phd/IrO2/structures/slab/result.txt", ' ');
```

```julia
x = data[:,1]
y = data[:,2]#
z = data[:,end-1].+7.037370210224886*data[:,3]#.-3.682801449591682;
number = data[:,3]

xvals = unique(sort(x))
yvals = unique(sort(y))
nvals = unique(sort(number))

nx, ny = length(xvals), length(yvals)
```

```julia
total = reshape(z, (ny, nx))

surface(xvals, yvals, total)

xlabel!("slab thickness in ang")
ylabel!("vacuum thickness in ang")
zlabel!("Surface Energy")
title!("Surface Energy for IrO2 (110) surface")
#zlims!(3.6,3.95)
```

```julia
matrix = [nvals ones(length(nvals))]
```
